```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
data = pd.read_csv('amazon_seller_info.csv')
data
```

```python
# How many rows and columns are present in the dataset?

data.shape
```

```python
# What are the data types of each column?

data.dtypes
```

```python
# Are there any missing values in the dataset?

data.isnull().sum()
```

```python
# Are there any duplicate records in the dataset?

data.duplicated().sum()
```

```python
# What is the distribution of sellers across different countries?

country_count = data['country'].value_counts()

plt.bar(country_count.index , country_count.values, color = "Yellow")
plt.xlabel('countries')
plt.ylabel('No of sellers')
plt.title('no of sellers in countries')
plt.show()
```

```python
# Which product category has the highest number of sellers?

category_count =  data['product_category'].value_counts()
category_count.head(1)
```

```python
# What is the average seller rating?

np.average(data['seller_rating'])
```

```python
data['seller_rating'].mean()
```

```python
# What is the distribution of sellers based on fulfillment method?

fulfillment_method_count = data['fulfillment_method'].value_counts()
plt.pie(fulfillment_method_count, labels= fulfillment_method_count.index, autopct = '%1.1f%%')
plt.show()
```

```python
# Which seller has the highest number of products?

data[['seller_name','total_products']].sort_values(by = 'total_products', ascending=False).head(1)
```

```python
# What is the average product price for each product category?

data[['avg_product_price','product_category']].groupby('product_category').mean()
```

```python
# Which product category has the highest average product price?

data[['avg_product_price','product_category']].groupby('product_category').mean().sort_values(by='avg_product_price', ascending = False).head(1)
```

```python
# Which country has the highest average seller rating?

data[['country','seller_rating']].groupby('country').mean().sort_values(by = 'seller_rating', ascending = False).head(1)
```

```python
# What are the top 5 sellers based on review count?

data[['seller_name', 'review_count']].sort_values(by = 'review_count', ascending = False).head(5)
```

```python
# What is the distribution of sellers based on their years of activity ?

count_of_activity_years = data['years_active'].value_counts()
plt.bar(count_of_activity_years.index, count_of_activity_years.values, color= 'Black')
plt.xlabel('Years of activity')
plt.ylabel('No of sellers')
plt.title('distribution of sellers based on their years of activity')
plt.show()
```
